a) Given the formula for log entropy describe what the purpose of the first component is (the local weight logarithm)  $$\log(1 + f_{ij})$$

Its purpouse is to dampen the weights for very frequent terms. The $1$ in $\log(1 + f_{ij})$ is used to also include terms with no frequency.

b) Given the formula for Log-Entropy describe what the purpose of the second component is (the entropy calculation).  $$ \left[1 + \left (\sum_{j} \frac{p_{ij} \log(p_{ij})}{\log(n)} \right)\right]$$

Its purpose is to calculate the entropy of each term, how spread out term $i$ is across the documents $j$ . The sum $\sum_{j} p_{ij} \log(p_{ij})$ calculates the actual entropy. Dividing by $\log(n)$ gives the normalized entropy($\log(n)$ is the maximum possible entropy a term can have across n documents). Because this value is negative, we add $1$ to it.

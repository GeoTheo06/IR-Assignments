# Assignment 4: Comparison Metrics

The four comparison operators are Euclidean distance, Pearson correlation, Spearman correlation and Manhattan distance. Each is tested on the Shakespeare collection with TF, TF-IDF and Log-Entropy using the query "scotland kings and thanes".

The implementations are in `inverted_index.py`. The `search()` selects the operator through its `comparison` argument. It sorts Euclidean and Manhattan distances from smallest to largest, and Pearson and Spearman coefficients from largest to smallest.

```python
import nltk

nltk.data.path.insert(0, "./nltk_data")
try:
    nltk.data.find("corpora/stopwords")
except LookupError:
    nltk.download("stopwords", download_dir="./nltk_data", raise_on_error=True)

from inverted_index import InvertedIndex
from utils import read_data
from part_b import run_experiment, WEIGHTINGS

index = InvertedIndex()
for document in read_data("./shakespeare"):
    index.add_document(document)
```

## 1. Euclidean distance

$$d(q,v)=\sqrt{\sum_i(q_i-v_i)^2}$$

`euclidean_comparison()` adds the squared differences between corresponding entries and returns their square root. Smaller distances rank first. Each experiment builds the document matrix and query with the same weighting scheme.

```python
for weighting in WEIGHTINGS:
    run_experiment(index, weighting, "euclidean")
```

**Evaluation:** Macbeth ranks 7th with TF, 27th with TF-IDF and 11th with Log-Entropy. `A Lover's Complaint` ranks first in all three runs. These results are less correct for a query about Scotland, kings and thanes. The comparison uses the full vocabulary without vector normalisation so document magnitude strongly affects the distance. When terms are present in a document but absent from the short query contribute to the distance, which will prefer works with smaller vectors.

## 2. Pearson correlation

`pearson_comparison()` returns the coefficient between the query and document vectors. Larger coefficients rank first.

```python
for weighting in WEIGHTINGS:
    run_experiment(index, weighting, "pearson")
```

**Evaluation:** With TF, `King Henry V` ranks first and Macbeth ranks 12th. With both TF-IDF and Log-Entropy, Macbeth ranks first. Raw term counts give strong influence to the common term `king`. Weighting makes distinctive query terms more influential, producing a more correct result. Pearson therefore works better with TF-IDF or Log-Entropy than with TF for this query.

## 3. Spearman correlation

`spearman_comparison()` returns its correlation coefficient. It compares the ranks of term weights within each vector. Larger coefficients rank first.

```python
for weighting in WEIGHTINGS:
    run_experiment(index, weighting, "spearman")
```

**Evaluation:** Macbeth ranks first with TF, TF-IDF and Log-Entropy. Comparing ranks reduces the influence of unusually large term counts, so the raw frequency of `king` has less influence than in the TF Pearson experiment. Spearman gives a correct first result under all three schemes for this query. However, many positions have zero values, and rank comparison discards information about the size of weight differences.

## 4. Manhattan distance

$$d(q,v)=\sum_i|q_i-v_i|$$

`manhattan_comparison()` adds the absolute differences between corresponding entries. Smaller distances rank first.

Reference: [SciPy's city-block (Manhattan) distance documentation](https://docs.scipy.org/doc/scipy/reference/generated/scipy.spatial.distance.cityblock.html). The implementation uses a loop that calculates that formula.

```python
for weighting in WEIGHTINGS:
    run_experiment(index, weighting, "manhattan")
```

**Evaluation:** `A Lover's Complaint` ranks first under all three schemes. Macbeth ranks 9th with TF, 13th with TF-IDF and 16th with Log-Entropy. Manhattan is less correct for this query because document terms absent from the query still contribute their weights to the distance, and total document weight strongly influences the ranking.

For this query, Spearman with all three schemes and Pearson with TF-IDF or Log-Entropy place Macbeth first. The distance methods give less correct rankings. This comparison however may have turned out differently with different queries.
